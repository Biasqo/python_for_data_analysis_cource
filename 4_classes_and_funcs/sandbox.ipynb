```python
from typeguard import typechecked
```

```python
def square(n: int) -> int:
    return n * n
```

```python
square(5.5)
```

```python
@typechecked
def square_typed(n: int) -> int:
    return n * n
```

```python
square_typed(5.5)
```

```python
%%time
square_typed("int")
```

```python
%%time
square("int")
```

```python
class BankAccount:
    def __init__(self, account_holder_name: str, balance: float):
        self.account_holder_name = account_holder_name
        self.balance = balance

    def calc_money(self, to_cut: int) -> float:
        try:
            self.balance /= to_cut
        except ZeroDivisionError as e:
            raise BankAccountException("Something went wrong") from None

    def withdraw_from_account(self, amount: float):
        if (self.balance - amount) < 0:
            raise NegativeBankAccountException("Денег нет :(")
        self.balance -= amount

    def add_money_to_balance(self, amount: float):
        if (self.balance + amount) >= 2_000_000_000:
            raise TooMuchMoneyException("Много денег слишком - делись!")
        self.balance += amount
```

```python
andrey_bank_account = BankAccount(account_holder_name='Andrew', balance=1000)
```

```python
andrey_bank_account.add_money_to_balance(2_000_000_000)
```

```python
andrey_bank_account.balance
```

```python
class BankAccountException(Exception):
    pass
```

```python
class NegativeBankAccountException(BankAccountException):
    pass
```

```python
class TooMuchMoneyException(BankAccountException):
    pass
```
