#### F-строки

```python
product = "laptop"
units = 3
price_per_unit = 999.99
```

```python
# Простые операции
print(f"Sold {units} {product}s")

# Форматирование
print(f"Total: ${units * price_per_unit:,.2f}")
print(f"Tax (20%): ${units * price_per_unit * 0.2:.2f}")

# Простой дебаг через принт
print(f"{units=}, {price_per_unit=}")
```

#### Функции

```python
def classify_gdp_growth(rate: float) -> str:
    if rate < 0:
        return "recession"
    elif rate < 2:
        return "slow"
    elif rate < 4:
        return "moderate"
    else:
        return "high"
```

```python
print(classify_gdp_growth(3.1))
```

```python
def add_transaction(amount, history=[]):
    history.append(amount)
    return history
```

```python
print(add_transaction(100))
print(add_transaction(200))
```

```python
def add_transaction(amount, history=None):
    if history is None:
        history = []
    history.append(amount)
    return history
```

```python
print(add_transaction(100))
print(add_transaction(200))
```
