#### Приведение типов в Python

| Преобразование               | Функция / Пример              | Потеря данных / Риск                                |
|------------------------------|-------------------------------|-----------------------------------------------------|
| `float` → `int`              | `int(3.9)` → `3`              | Дробная часть отбрасывается (теряется)              |
| `int` → `float` (большое)    | `float(10**30)` → `1e+30`     | Потеря точности (ограниченная мантисса)             |
| `str` → `int` / `float`      | `int("42")`                   | `ValueError`, если строка не число                   |
| `bool` → `int`               | `int(True)` → `1`             | Потери нет (но смысл значения исчезает)             |
| `int` / `float` → `bool`     | `bool(0)` → `False`           | Значение сводится к истинности (0/не-0)             |

### Изменяемые и неизменяемые типы

| Категория | Тип | Изменяем? | Пример |
|----------|-------|-------------|---------|
| **MUTABLE** | `list` | ✅ Yes | `lst[0] = 99` |
| **MUTABLE** | `dict` | ✅ Yes | `d["key"] = "value"` |
| **MUTABLE** | `set` | ✅ Yes | `s.add(4)` |
| **MUTABLE** | Custom objects | ✅ Yes | `obj.attr = 10` |
| **IMMUTABLE** | `str` | ❌ No | `s[0] = "H"` → ERROR |
| **IMMUTABLE** | `int` | ❌ No | `x = x + 1` creates NEW |
| **IMMUTABLE** | `float` | ❌ No | `f = f * 2` creates NEW |
| **IMMUTABLE** | `tuple` | ❌ No | `t[0] = 99` → ERROR |
| **IMMUTABLE** | `bool` | ❌ No | Can't change True/False |
| **IMMUTABLE** | `frozenset` | ❌ No | No add/remove methods |
| **IMMUTABLE** | `None` | ❌ No | Always the same |

#### F-строки

```python
product = "laptop"
units = 3
price_per_unit = 999.99
```

```python
# Простые операции
print(f"Sold {units} {product}s")

# Форматирование
print(f"Total: ${units * price_per_unit:,.2f}")
print(f"Tax (20%): ${units * price_per_unit * 0.2:.2f}")

# Простой дебаг через принт
print(f"{units=}, {price_per_unit=}")
```

#### Операторы

##### Арифметические
| Оператор | Что делает | Пример |
|----------|------------|--------|
| `+` | Сложение | `5 + 3` → `8` |
| `-` | Вычитание | `5 - 3` → `2` |
| `*` | Умножение | `5 * 3` → `15` |
| `/` | Деление (float) | `5 / 2` → `2.5` |
| `//` | Целочисленное деление | `5 // 2` → `2` |
| `%` | Остаток | `5 % 2` → `1` |
| `**` | Степень | `5 ** 2` → `25` |

##### Сравнения
| Оператор | Что делает | Пример |
|----------|------------|--------|
| `==` | Равно | `5 == 3` → `False` |
| `!=` | Не равно | `5 != 3` → `True` |
| `>` | Больше | `5 > 3` → `True` |
| `<` | Меньше | `5 < 3` → `False` |
| `>=` | Больше или равно | `5 >= 5` → `True` |
| `<=` | Меньше или равно | `5 <= 3` → `False` |

##### Логические
| Оператор | Что делает | Пример |
|----------|------------|--------|
| `and` | И | `True and False` → `False` |
| `or` | Или | `True or False` → `True` |
| `not` | Не | `not True` → `False` |

##### Присваивания
| Оператор | Что делает | Пример |
|----------|------------|--------|
| `=` | Присвоить | `x = 5` |
| `+=` | Прибавить | `x += 3` → `x = x + 3` |
| `-=` | Вычесть | `x -= 3` → `x = x - 3` |
| `*=` | Умножить | `x *= 3` → `x = x * 3` |
| `/=` | Разделить | `x /= 3` → `x = x / 3` |
| `//=` | Целочисл. деление | `x //= 3` |
| `%=` | Остаток | `x %= 3` |
| `**=` | Степень | `x **= 3` |

##### Работа с объектами
| Оператор | Что делает | Пример |
|----------|------------|--------|
| `is` | Один объект? | `a is b` |
| `is not` | Разные объекты? | `a is not b` |
| `in` | Входит в? | `1 in [1, 2]` → `True` |
| `not in` | Не входит? | `3 not in [1, 2]` → `True` |

##### Битовые
| Оператор | Что делает | Пример |
|----------|------------|--------|
| `&` | AND | `5 & 3` → `1` |
| `\|` | OR | `5 \| 3` → `7` |
| `^` | XOR | `5 ^ 3` → `6` |
| `~` | NOT | `~5` → `-6` |
| `<<` | Сдвиг влево | `5 << 1` → `10` |
| `>>` | Сдвиг вправо | `5 >> 1` → `2` |

##### Приоритет (от высокого к низкому)
| Приоритет | Операторы |
|-----------|-----------|
| 1 (высший) | `**` |
| 2 | `*` `/` `//` `%` |
| 3 | `+` `-` |
| 4 | `==` `!=` `>` `<` `>=` `<=` |
| 5 | `not` |
| 6 | `and` |
| 7 | `or` |
| 8 (низший) | `=` `+=` `-=` и др. |

#### Условные операторы

```python
# Простой if
age = 18
if age >= 18:
    print('Взрослый')
```

```python
# if-else
score = 75
if score >= 60:
    print('Сдал')
else:
    print('Не сдал')
```

```python
# if-elif-else
grade = 85
if grade >= 90:
    print('A')
elif grade >= 80:
    print('B')
elif grade >= 70:
    print('C')
else:
    print('D')
```

#### Циклы

##### While

```python
# Счётчик
i = 0
while i < 5:
    print(i)
    i += 1
```

```python
# Бесконечный (выход через break)
while True: # антипаттерн
    user = input('Введи exit: ')
    if user == 'exit':
        break
```

```python
# С else (выполняется если не было break)
i = 0
while i < 3:
    print(i)
    i += 1
else:
    print('Цикл завершён')
```

##### For

```python
# По списку
for item in ['a', 'b', 'c']:
    print(item)
```

```python
# По диапазону
for i in range(5):
    print(i)
```

```python
# По строке
for char in 'hello':
    print(char)
```

```python
# С индексом
for i, item in enumerate(['x', 'y', 'z']):
    print(i, item)
```

#### Функции

```python
def classify_gdp_growth(rate: float) -> str:
    if rate < 0:
        return "recession"
    elif rate < 2:
        return "slow"
    elif rate < 4:
        return "moderate"
    else:
        return "high"
```

```python
print(classify_gdp_growth(3.1))
```

```python
# enumerate()
# Добавляет счётчик к итерации. Возвращает `(индекс, значение)`.
countries = ["Germany", "France", "Italy"]
for i, country in enumerate(countries, start=1):
    print(f"{i}. {country}")
```

```python
def add_transaction(amount, history=[]):
    history.append(amount)
    return history
```

```python
print(add_transaction(100))
print(add_transaction(200)) # это верный вывод?
```
